```python
import json
import numpy as np
from shapely.geometry import shape, Polygon
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm
from PIL import Image, ImageDraw
import os
# Function to load GeoJSON data from a file
def load_geojson_data(file_path):
    with open(file_path, 'r') as f:
        geojson_data = json.load(f)
    return geojson_data

# Function to invert the CDL_CLASSES dictionary
def invert_class_dict(class_dict):
    return {v: k for k, v in class_dict.items()}

# Function to compute affine transformation parameters
def compute_affine_transform(roi_polygon, image_size):
    minx, miny, maxx, maxy = roi_polygon.bounds
    x_range = maxx - minx
    y_range = maxy - miny
    image_width, image_height = image_size
    transform_params = {
        'minx': minx,
        'miny': miny,
        'maxy': maxy,  # Added maxy to the params
        'x_range': x_range,
        'y_range': y_range,
        'image_width': image_width,
        'image_height': image_height
    }
    return transform_params

# Function to transform world coordinates to image coordinates
def world_to_image_coords(x, y, transform_params):
    minx = transform_params['minx']
    maxy = transform_params['maxy']
    x_range = transform_params['x_range']
    y_range = transform_params['y_range']
    image_width = transform_params['image_width']
    image_height = transform_params['image_height']
    ix = int((x - minx) / x_range * (image_width - 1))
    iy = int((maxy - y) / y_range * (image_height - 1))  # Invert y-axis for image coordinates
    return ix, iy

# Updated rasterize_features function
def rasterize_features(geojson_data, roi_polygon, label_to_int, transform_params, image_size):
    image_width, image_height = image_size
    # Create an empty image for drawing
    mask_image = Image.new('L', (image_width, image_height), 0)
    draw = ImageDraw.Draw(mask_image)

    # Iterate over each feature in the GeoJSON data
    for feature in geojson_data['features']:
        label = feature['properties']['subclass_name_crop2']
        geometry = feature['geometry']
        feature_shape = shape(geometry)

        # Check if the feature overlaps with the ROI
        if feature_shape.intersects(roi_polygon):
            # Calculate the overlapping area
            intersection = feature_shape.intersection(roi_polygon)
            if intersection.is_empty:
                continue

            # Get the integer value for the label
            value = label_to_int.get(label, 0)  # Default to 0 (Unknown) if label not found

            # Rasterize the geometry onto the mask image
            if intersection.geom_type == 'Polygon':
                polygons = [intersection]
            elif intersection.geom_type == 'MultiPolygon':
                polygons = list(intersection.geoms)
            else:
                continue  # Skip other geometry types

            for poly in polygons:
                # Get exterior coordinates (extracting only x and y)
                exterior_coords = [(coord[0], coord[1]) for coord in poly.exterior.coords]
                # Transform to image coordinates
                exterior_pixels = [world_to_image_coords(x, y, transform_params) for x, y in exterior_coords]

                # Get interior coordinates (holes)
                interiors = []
                for interior in poly.interiors:
                    interior_coords = [(coord[0], coord[1]) for coord in interior.coords]
                    interior_pixels = [world_to_image_coords(x, y, transform_params) for x, y in interior_coords]
                    interiors.append(interior_pixels)

                # Draw the polygon onto the mask image
                draw.polygon(exterior_pixels, fill=value)
                for interior in interiors:
                    draw.polygon(interior, fill=0)  # Assuming holes should be filled with 0

    return mask_image


def display_and_save_mask_with_class_names(mask_image, class_dict, cmap='nipy_spectral',
                                           filename='/Users/maral/Desktop/satelite_img/out1/roi_mask.tiff', image_size=None):
                                           
    # Convert the mask image to a numpy array
    mask = np.array(mask_image)

    # Identify unique class values in the mask
    unique_values = np.unique(mask)

    # Create a mapping from class values to class names
    value_to_label = {value: class_dict.get(value, 'Unknown') for value in unique_values}

    # Create a colormap and norm that only include the unique values
    cmap = plt.get_cmap(cmap, len(unique_values))
    norm = BoundaryNorm(boundaries=np.arange(len(unique_values)+1)-0.5, ncolors=len(unique_values))

    # Map the mask values to indices in the unique values array
    value_to_index = {value: idx for idx, value in enumerate(unique_values)}
    indexed_mask = np.vectorize(value_to_index.get)(mask)

    # Display the mask image using matplotlib
    plt.figure(figsize=(10, 8))
    im = plt.imshow(indexed_mask, cmap=cmap, norm=norm)
    plt.title('ROI Mask with Class Labels')

    # Create custom ticks and labels for the colorbar
    cbar = plt.colorbar(im, ticks=range(len(unique_values)))
    cbar_labels = [value_to_label[value] for value in unique_values]
    cbar.ax.set_yticklabels(cbar_labels)
    cbar.ax.tick_params(labelsize=8)
    cbar.set_label('Class Name', rotation=270, labelpad=15)

    plt.show()

    # Resize the mask image to the desired size
    mask_image = mask_image.resize(image_size, resample=Image.BICUBIC)

    # Ensure the directory exists
    os.makedirs(os.path.dirname(filename), exist_ok=True)

    # Save the mask image in TIFF format
    mask_image.save(filename)

    print(f"Mask image saved as {filename}")

```

```python
import os
import openai
from moviepy.editor import VideoFileClip

def extract_audio(video_path, audio_path):
    """
    Extracts audio from a video file and saves it as an mp3 file.
    
    :param video_path: Path to the input video file.
    :param audio_path: Path to save the extracted audio file.
    """
    try:
        video = VideoFileClip(video_path)
        video.audio.write_audiofile(audio_path, codec='mp3')
        video.close()
        print(f"Audio extracted and saved to {audio_path}")
    except Exception as e:
        print(f"Error extracting audio: {e}")
        raise

def transcribe_audio(audio_path, api_key, language="en"):
    """
    Transcribes audio using OpenAI's Whisper API.
    
    :param audio_path: Path to the audio file to transcribe.
    :param api_key: OpenAI API key.
    :param language: Language of the audio.
    :return: Transcribed text.
    """
    openai.api_key = api_key
    try:
        with open(audio_path, "rb") as audio_file:
            print("Uploading audio for transcription...")
            transcript = openai.Audio.transcribe("whisper-1", audio_file, language=language)
            print("Transcription completed.")
            return transcript["text"]
    except Exception as e:
        print(f"Error during transcription: {e}")
        raise

def save_transcript(transcript, output_path):
    """
    Saves the transcript to a text file.
    
    :param transcript: The transcribed text.
    :param output_path: Path to save the transcript.
    """
    try:
        with open(output_path, "w", encoding="utf-8") as file:
            file.write(transcript)
        print(f"Transcript saved to {output_path}")
    except Exception as e:
        print(f"Error saving transcript: {e}")
        raise

def main():
    # ---------------------- User Configuration ----------------------
    video_file = "input_video.mp4"      # Path to your input .mp4 file
    audio_file = "extracted_audio.mp3"  # Temporary audio file
    transcript_file = "transcript.txt"  # Output transcript file
    api_key = "YOUR_OPENAI_API_KEY"      # Replace with your OpenAI API key
    # ----------------------------------------------------------------

    # Check if the video file exists
    if not os.path.isfile(video_file):
        print(f"Video file {video_file} does not exist.")
        return

    try:
        # Step 1: Extract audio from video
        extract_audio(video_file, audio_file)

        # Step 2: Transcribe audio to text
        transcript = transcribe_audio(audio_file, api_key, language="en")

        # Step 3: Save the transcript to a text file
        save_transcript(transcript, transcript_file)

    finally:
        # Clean up the extracted audio file if it exists
        if os.path.exists(audio_file):
            os.remove(audio_file)
            print(f"Temporary audio file {audio_file} removed.")

if __name__ == "__main__":
    main()
```
