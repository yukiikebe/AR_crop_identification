```python
import glob
import numpy as np
import tifffile

# Iterate over all .tif files in the current directory
for filename in glob.glob("/Users/maral/Desktop/satelite_img/out1/*.tiff"):
    # Read the image
    img = tifffile.imread(filename)
    
    # Get the unique pixel values
    unique_vals = np.unique(img)
    
    # Check the range of unique values
    if unique_vals.min() >= 0 and unique_vals.max() <= 53:
        print(f"{filename}: Unique values are within{unique_vals}.")
    else:
        print(f"{filename} {unique_vals}: no")
```
