```python
from PIL import Image, ImageDraw, ImageFont
import random

# Constants
grid_size = 20
num_grids = 20
image_size = grid_size * num_grids
margin = 30  # Margin to accommodate numbers outside the grid
total_size = image_size + margin

colors = [(0, 0, 255), (0, 255, 0)]  # Blue and Green
color_distribution = [0] * 20 + [1] * 80  # 20% blue, 80% green

# Create a new image with white background
image = Image.new('RGB', (total_size, total_size), (255, 255, 255))
draw = ImageDraw.Draw(image)

# Load a font
font = ImageFont.load_default()

# Draw the grids
for x in range(margin, margin + image_size, grid_size):
    for y in range(margin, margin + image_size, grid_size):
        color_index = random.choice(color_distribution)
        color = colors[color_index]
        draw.rectangle([x, y, x + grid_size, y + grid_size], fill=color)

# Draw the lines between the grids
for x in range(margin, margin + image_size + 1, grid_size):
    draw.line((x, margin, x, margin + image_size), fill=(0, 0, 0))  # Vertical lines
for y in range(margin, margin + image_size + 1, grid_size):
    draw.line((margin, y, margin + image_size, y), fill=(0, 0, 0))  # Horizontal lines

# Draw the numbers along the x-axis and y-axis outside the grid
for i in range(1, num_grids + 1):
    draw.text((margin + (i - 1) * grid_size + grid_size // 2, margin // 2), str(i), fill=(0, 0, 0), font=font, anchor="mm")  # x-axis
    draw.text((margin // 2, margin + (i - 1) * grid_size + grid_size // 2), str(i), fill=(0, 0, 0), font=font, anchor="mm")  # y-axis

# Save the image
image.save('grid_image.png')
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import rasterio

def save_tci_image(blue_band_path, green_band_path, red_band_path, output_path):
    # Open the TIF files
    if not (os.path.isfile(blue_band_path) and os.path.isfile(green_band_path) and os.path.isfile(red_band_path)):
        return

    with rasterio.open(blue_band_path) as blue_band:
        blue = blue_band.read(1)
    with rasterio.open(green_band_path) as green_band:
        green = green_band.read(1)
    with rasterio.open(red_band_path) as red_band:
        red = red_band.read(1)

    # Stack the bands
    stacked_data = np.stack((red, green, blue), axis=-1)
    plt.imsave(output_path, stacked_data)

```
