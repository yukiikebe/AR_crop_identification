```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt
# Read the CDL image
cdl_image_path = "/home/vuonghn/research/dataset/satellite/arkansas/org_maral/cdl/rgb_2023-06-10_cdl.tif"

with rasterio.open(cdl_image_path) as src:
    cdl_image = src.read(1)  # Read the first band
unique_values = np.unique(cdl_image)
value_mapping = {val: idx for idx, val in enumerate(unique_values)}
cdl_image_normalized = np.vectorize(value_mapping.get)(cdl_image)

print("cdl_image ", np.unique(cdl_image))
print("cdl_image_normalized ", np.unique(cdl_image_normalized))

normalized_image_path = "/home/vuonghn/research/dataset/satellite/arkansas/org_maral/cdl/rgb_27classes_cdl.tif"
with rasterio.open(cdl_image_path) as src:
    metadata = src.meta.copy()
metadata.update(dtype=rasterio.uint8)
# Write the normalized image to a new TIFF file
with rasterio.open(normalized_image_path, 'w', **metadata) as dst:
    dst.write(cdl_image_normalized.astype(rasterio.uint8), 1)

print(f"Normalized image saved to {normalized_image_path}")
```
